```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../1_datasets/cleaned_data.csv")
df.head()
```

```python
df.columns
```

```python
# choose the columns to explore
columns = [
    "Project_Type",
    "Country",
    "Average_Site_Size(m²)",
    "Project_Duration(Months)",
    "Primary_Energy_Source",
    "Average_Daily_Energy_Consumption(KWh)",
    "Peak_Power_Demand(kW)",
    "Daily_Generator_Runtime(Hours)",
    "Meter_By_Equipment",
    "Generator_Sizes(KVA)",
    "Diesel_Consumption_Per_Week(Liters)",
    "Diesel_Delivery_Per_Week",
    "Considered_Renewables",
    "Space_For_Solar",
    "Consider_Hybrid_System",
    "Track_CO2_Emissions",
    "Sustainability_Requirements",
    "Willing_To_Share_Data",
    "Join_Pilot_Study",
    "Industry_Role",
]
# Create count plots for each categorical column
for col in columns:
    plt.figure(figsize=(10, 6))  # Adjust figure size for better readability
    # Use plt.pie for pie charts
    counts = df[col].value_counts()
    plt.pie(counts, labels=counts.index, autopct="%1.1f%%", startangle=140)
    plt.title(f"Distribution of {col}")
    plt.axis("equal")  # Equal aspect ratio ensures that pie is drawn as a circle.
    plt.tight_layout()
    plt.show()
```

```python
# Grouped bar plot for Project_Type and Primary_Energy_Source
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x="Project_Type", hue="Primary_Energy_Source")
plt.title("Project Type by Primary Energy Source")
plt.xlabel("Project Type")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Grouped bar plot for Country and Primary_Energy_Source
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x="Country", hue="Primary_Energy_Source")
plt.title("Country by Primary Energy Source")
plt.xlabel("Country")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
from collections import Counter

# Flatten all values into a single list
all_items = []
for row in df["Renewable_Concerns"].dropna():
    items = [x.strip() for x in row.split(",")]
    all_items.extend(items)

# Count frequency using Counter
freq = Counter(all_items)
# Convert to DataFrame for plotting
freq_df = pd.DataFrame(freq.items(), columns=["Item", "Count"]).sort_values(
    "Count", ascending=False
)

# Plot
plt.figure(figsize=(10, 6))
plt.bar(freq_df["Item"], freq_df["Count"], color="skyblue")
plt.title("Frequency of Renewable Concerns")
plt.xlabel("Renewable Concerns")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
all_items = []
for row in df["Power_Source_Factors"].dropna():
    items = [x.strip() for x in row.split(",")]
    all_items.extend(items)

# Count frequency using Counter
freq = Counter(all_items)
# Convert to DataFrame for plotting
freq_df = pd.DataFrame(freq.items(), columns=["Item", "Count"]).sort_values(
    "Count", ascending=False
)

# Plot
plt.figure(figsize=(10, 6))
plt.bar(freq_df["Item"], freq_df["Count"], color="skyblue")
plt.title("Frequency of Power Source Factors")
plt.xlabel("Power Source Factors")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df1 = df.copy()
priority_order = ["< 50", "50 - 150", "150 - 300", "> 300", "Not applicable"]

# Convert to ordered categorical type
df1["Generator_Sizes(KVA)"] = pd.Categorical(
    df["Generator_Sizes(KVA)"], categories=priority_order, ordered=True
)

# Now sort
df1_sorted = df1.sort_values("Generator_Sizes(KVA)")
```

```python
df2 = df.copy()
priority_order = ["< 6", "6 - 12", "12 - 24", "> 24"]

# Convert to ordered categorical type
df2["Project_Duration(Months)"] = pd.Categorical(
    df2["Project_Duration(Months)"], categories=priority_order, ordered=True
)

# Now sort
df2_sorted = df2.sort_values("Project_Duration(Months)")
```

```python
# Cross-tabulate
ct = pd.crosstab(
    df2_sorted["Project_Duration(Months)"], df1_sorted["Generator_Sizes(KVA)"]
)

# Heatmap
sns.heatmap(ct, annot=True, cmap="Blues")
plt.title("Heatmap of Project Duration vs Generator Sizes")
plt.xlabel("Generator Sizes")
plt.ylabel("Project Duration")
plt.show()
```

```python
df3 = df.copy()
priority_order = ["< 50", "50 - 200", "200 - 500", "> 500"]

# Convert to ordered categorical type
df3["Average_Daily_Energy_Consumption(KWh)"] = pd.Categorical(
    df3["Average_Daily_Energy_Consumption(KWh)"],
    categories=priority_order,
    ordered=True,
)

# Now sort
df3_sorted = df3.sort_values("Average_Daily_Energy_Consumption(KWh)")
df3_sorted["Average_Daily_Energy_Consumption(KWh)"].value_counts()
```

```python
# Cross-tabulate
ct = pd.crosstab(
    df3_sorted["Average_Daily_Energy_Consumption(KWh)"],
    df1_sorted["Generator_Sizes(KVA)"],
)

# Heatmap
sns.heatmap(ct, annot=True, cmap="Blues")
plt.title("Heatmap of Project Duration vs Generator Sizes")
plt.xlabel("Generator Sizes")
plt.ylabel("Average_Daily_Energy_Consumption(KWh)")
plt.show()
```
