# Data Exploration

This notebook contains exploratory data analysis (EDA) of the merged dataset.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../2_data_preparation/merged_projects_plus_costs.csv')
df.head()
```

```
print("Shape:", df.shape)
print("\nData types:\n", df.dtypes)
print("\nMissing values (%):\n", (df.isna().mean()*100).round(1))
```

```
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Project_Type", order=df["Project_Type"].value_counts().index)
plt.xticks(rotation=45, ha="right")
plt.title("Distribution of Project Types")
plt.show()
```

```
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Primary_Source_of_Energy_Used_on_your_construction_site",
              order=df["Primary_Source_of_Energy_Used_on_your_construction_site"].value_counts().index)
plt.xticks(rotation=45, ha="right")
plt.title("Primary Energy Sources")
plt.show()
```

```
plt.figure(figsize=(7,4))
sns.boxplot(data=df, x="Country", y="Monthly_Generator_Maintenance_Cost_AED")
plt.xticks(rotation=45, ha="right")
plt.title("Monthly Generator Maintenance Cost (AED) by Country")
plt.show()
```

```
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Daily_Generator_Runtime", order=df["Daily_Generator_Runtime"].value_counts().index)
plt.xticks(rotation=45, ha="right")
plt.title("Daily Generator Runtime Distribution")
plt.show()
```
