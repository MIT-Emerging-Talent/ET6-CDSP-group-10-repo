```python
def clean_column_name(name):
    name = name.strip() # Remove leading/trailing whitespace
    name = name.replace(" (", "(")  # Remove space before (
    name = name.replace("  ", " ")  # Handle double spaces if any
    name = name.replace("Daily_Generator_Runtime (Hours)", "Daily_Generator_Runtime(Hours)")
    return name

# Apply to all column names
df.columns = [clean_column_name(col) for col in df.columns]

df.columns
```

```python
def simplify_gen_issues(text):
    if pd.isna(text):
        return text

    parts = [p.strip() for p in text.split(',')]
    labels = []

    if any("High fuel cost and fuel logistics" in p for p in parts):
        labels.append("Fuel Cost & Logistics")
    if any("Frequent maintenance" in p for p in parts):
        labels.append("Maintenance")
    if any("Long idle time" in p for p in parts):
        labels.append("Idle Time")

    return " + ".join(labels)

# Apply to original column
df['Common_Gen_Issues'] = df['Common_Gen_Issues'].apply(simplify_gen_issues)
df['Common_Gen_Issues'].value_counts()
```
